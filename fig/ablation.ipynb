```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# 设置 Seaborn 风格
sns.set_theme(style="whitegrid")
plt.rcParams['font.family'] = 'Optima'
plt.rcParams['font.weight'] = 'bold'
# plt.rcParams['font.size'] = 16
# Nature 杂志风格的配色方案
colors = sns.color_palette("muted")

# 模拟数据
x = [0,1, 3, 5, 10]  # 只有四个数据点
y_data = [
    [81, 79, 85, 89],  # 第一组数据
    [20, 30, 25, 35],  # 第二组数据
    [15, 25, 20, 30],  # 第三组数据
    [10, 20, 15, 25]   # 第四组数据
]

#${{N_\alpha}}$
data1 = [[[1, 81], [3, 79], [5, 85], [10, 89]],
           [[1, 85], [3, 85], [5, 86], [10, 88]]]
# data1_prenode = [[[1,40],[3,53],[5,60]]]
#middle node$
# data2 = [[[1, 85], ,
#            [[1, 86], ]]
data2 = [[[1, 85], [3, 91], [5, 94]],
           [[1, 86], [3, 88],  [5, 92]]]
#$|V_{r,n}^{l}|$ enhance
data3= [[[0,72],[1, 78], [3, 83], [5, 85], [10, 86]],
           [[0,80], [1, 80], [3, 81], [5, 86], [10, 88]]]


#replication
data4= [[[1, 85], [3, 93], [5, 98], [10, 100]],
           [ [1, 86], [3, 92], [5, 98], [10, 100]]]


data_full = [data1, data2, data3, data4]
# 创建一个包含4个子图的图形
fig, axes = plt.subplots(1, 4, figsize=(16, 4), sharey=True)  # Use sharey=True for consistent y-axis

# 为每个子图设置不同的 x 轴标签和刻度标签
x_labels = [
    r'Number of $N_\alpha$ corpus generation for Direct Attack',
    r'Number of $|\upsilon_{r, n}^{t}|^{\prime}$ in Direct Attack',
    r'Number of $|V_{r,n}^{l}|$ in Indirect Attack',
    'Number of Text Replication'
]

x_tick_labels = [
    ['0','1', '3', '5', '10'],
    ['0','1', '3', '5', '10'],
    ['0','1', '3', '5', '10'],
    ['0','1', '3', '5', '10'],

]

# 自定义的图例标签
legend_labels = [
    r'$\text{Label 1}$',
    r'$\text{Label 2}$',
    r'$\text{Label 3}$',
    r'$\text{Label 4}$'
]

# 绘制每个子图
for i, ax in enumerate(axes):

    ax.plot([d[0] for d in data_full[i][0]], [d[1] for d in data_full[i][0]], marker='o', color=colors[0], label='Location')
    ax.plot([d[0] for d in data_full[i][1]], [d[1] for d in data_full[i][1]], marker='o', color=colors[1], label='MED')
   
    ax.set_xlabel(x_labels[i],fontweight='bold')  # 设置不同的 x 轴标签
    if i == 0:
        ax.set_ylabel('ASR',fontweight='bold')  # Ensure y-axis label is set for each subplot
    ax.set_xticks(x)
    ax.set_xticklabels(x_tick_labels[i])  # 设置不同的 x 轴刻度标签
    if i == 0:
        ax.legend()
    # 添加网格线
    ax.grid(True, linestyle='--')

    # 隐藏特定的边框
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_visible(True)  # 显示左边框

    # 保留 y 轴的刻度线
    ax.tick_params(axis='y', length=0)

# 调整布局
plt.tight_layout()
plt.savefig('ablation.pdf')

# 显示图形
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Use default Matplotlib style
plt.style.use('default')
plt.rcParams['font.family'] = 'Optima'
plt.rcParams['font.weight'] = 'normal'

# Define colors manually (similar to Seaborn's "muted" palette)
colors = ['#4878CF', '#6ACC65', '#D65F5F', '#B47CC7', '#C4AD66', '#77BEDB']

# Simulated data
x = [0, 1, 3, 5, 10]
y_data = [
    [81, 79, 85, 89],
    [20, 30, 25, 35],
    [15, 25, 20, 30],
    [10, 20, 15, 25]
]

data1 = [[[1, 81], [3, 79], [5, 85], [10, 89]],
         [[1, 85], [3, 85], [5, 86], [10, 88]]]

data2 = [[[1, 85], [3, 91], [5, 94]],
         [[1, 86], [3, 88], [5, 92]]]

data3 = [[[0, 45], [1, 65], [3, 74], [5, 85], [10, 86]],
         [[0, 38], [1, 63], [3, 75], [5, 86], [10, 88]]]

data4 = [[[1, 85], [3, 93], [5, 98], [10, 100]],
         [[1, 86], [3, 92], [5, 98], [10, 100]]]

data_full = [data1, data2, data3, data4]

# Create a figure with 4 subplots
fig, axes = plt.subplots(1, 4, figsize=(20, 5), sharey=True)

# Labels for each subplot
x_labels = [
    r'Number of $N_\alpha$ corpus generation for Direct Attack',
    r'Number of $|\upsilon_{r, n}^{t}|^{\prime}$ in Direct Attack',
    r'Number of $|V_{r,n}^{l}|$ in Indirect Attack',
    'Number of Text Replication'
]

x_tick_labels = [
    ['0', '1', '3', '5', '10'],
    ['0', '1', '3', '5', '10'],
    ['0', '1', '3', '5', '10'],
    ['0', '1', '3', '5', '10'],
]

# Legend labels
legend_labels = [
    r'$\text{Label 1}$',
    r'$\text{Label 2}$',
    r'$\text{Label 3}$',
    r'$\text{Label 4}$'
]

# Plot each subplot
for i, ax in enumerate(axes):
    ax.plot([d[0] for d in data_full[i][0]], [d[1] for d in data_full[i][0]], marker='o', color=colors[0], label='Location')
    ax.plot([d[0] for d in data_full[i][1]], [d[1] for d in data_full[i][1]], marker='o', color=colors[1], label='MED')

    ax.set_xlabel(x_labels[i])
    ax.set_ylabel('ASR')
    ax.set_xticks(x)
    ax.set_xticklabels(x_tick_labels[i])
    ax.legend()
    ax.grid(True, linestyle='--')

    # Hide specific borders
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_visible(True)

    # Keep y-axis ticks
    ax.tick_params(axis='y', length=0)

# Adjust layout
plt.tight_layout()

# Show plot
plt.show()
```
