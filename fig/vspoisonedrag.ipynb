```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set Seaborn style
sns.set_theme(style="whitegrid")
plt.rcParams['font.family'] = 'Optima'
plt.rcParams['font.weight'] = 'bold'
# plt.rcParams['font.size'] = 16
# Nature journal style color palette
colors = sns.color_palette("muted")

data2 = [
    [[167.9, 92.6], [503.7, 96.3], [839.5, 98.3]],
    [[223, 88.9], [669, 89.9], [1115, 92.3]],
    [[245, 87.8], [735, 89.2], [1225, 90.3]]
]

data3 = [
    [[823.9, 55.3]],
    [[721.5, 69.4]],
    [[654.6, 62.5]]
]

# Labels for the legend
legend_labels = [
    'Location', 'Med', 'Cyber', 
    'Location-PoisonedRAG', 'Med-PoisonedRAG', 'Cyber-PoisonedRAG'
]

# Create a single plot
fig, ax = plt.subplots(figsize=(5, 5))

# Plot data2
for i, dataset in enumerate(data2):
    x = [point[0] for point in dataset]
    y = [point[1] for point in dataset]
    ax.plot(x, y, marker='o', label=legend_labels[i], color=colors[i])

# Plot data3
for i, dataset in enumerate(data3):
    x = [point[0] for point in dataset]
    y = [point[1] for point in dataset]
    ax.plot(x, y, marker='s', label=legend_labels[i ], color=colors[i ])

# Customize the plot
ax.set_xlabel('Token Used Per Question')
ax.set_ylabel('ASR')
ax.set_title('Token Used between ours and PoisonedRAG')
ax.grid(True, linestyle='--')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(True)
ax.tick_params(axis='y', length=0)

# Add legend
ax.legend()

# Adjust layout
plt.tight_layout()
plt.savefig('vspoisonedrag.pdf')

# Show plot
plt.show()
```
